```python
files = dbutils.fs.ls('/Volumes/global_partner/bronze/global_partner')
current_timestamp = datetime.now(pytz.timezone('America/Chicago'))
write_mode = 'append'
for file in files:
    if file.name.endswith('.csv'):
        #print(file.name) #path
        tab_name = file.name.replace('.csv','').strip()
        df = spark.read.csv(file.path,header=True,inferSchema=False)
        df = df.withColumn('ingest_date',F.lit(current_timestamp))
        #print(df.head(5))global_partner_project.bronze.order_items
        df.write.mode(write_mode).saveAsTable(f'global_partner_project.bronze.{tab_name}')
```
